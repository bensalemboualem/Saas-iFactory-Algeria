## 1️⃣ Setup & Configuration

Installation automatique des dépendances Python.

```python
# Installation des bibliothèques nécessaires
import sys
import subprocess

def install_packages():
    """Installe les packages requis"""
    packages = [
        'matplotlib',
        'pandas',
        'plotly',
        'openpyxl',
        'kaleido',  # Pour export images Plotly
        'reportlab',  # Pour génération PDF
        'jinja2',  # Pour templates
    ]
    
    print("📦 Installation des dépendances...\n")
    for package in packages:
        try:
            __import__(package)
            print(f"✅ {package} déjà installé")
        except ImportError:
            print(f"⏳ Installation de {package}...")
            subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", package])
            print(f"✅ {package} installé")
    
    print("\n✅ Toutes les dépendances sont installées !\n")

install_packages()

# Imports
import matplotlib.pyplot as plt
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import plotly.figure_factory as ff
from plotly.subplots import make_subplots
from datetime import date, datetime, timedelta
from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment
import warnings
warnings.filterwarnings('ignore')

print("✅ Imports réussis ! Vous pouvez continuer.")
```

## 2️⃣ Analyses Financières

Calculs automatiques de ROI et visualisations budgétaires.

```python
# 2.1 - Calcul ROI par initiative

def calculate_roi(investment, revenue, label=""):
    """Calcule le ROI d'une initiative"""
    if investment == 0:
        roi = float('inf') if revenue > 0 else 0
        roi_display = "∞" if revenue > 0 else "0"
    else:
        roi = (revenue - investment) / investment
        roi_display = f"{roi:.1f}x"
    
    profit = revenue - investment
    
    print(f"📈 {label}")
    print(f"   Investissement: €{investment:,}")
    print(f"   Revenus:        €{revenue:,}")
    print(f"   Profit:         €{profit:,}")
    print(f"   ROI:            {roi_display}")
    print()
    
    return roi, profit

print("💰 Calcul ROI des Initiatives Clés 2026\n")
print("="*60)

# AWS Marketplace
roi_aws, profit_aws = calculate_roi(0, 200_000, "AWS Marketplace")

# n8n Embed License
roi_n8n, profit_n8n = calculate_roi(10_000, 100_000, "n8n Embed License")

# Prismatic iPaaS
roi_prismatic, profit_prismatic = calculate_roi(30_000, 500_000, "Prismatic Embedded iPaaS")

# GoHighLevel
roi_ghl, profit_ghl = calculate_roi(10_000, 84_000, "GoHighLevel Marketplace")

# RapidAPI
roi_rapid, profit_rapid = calculate_roi(0, 40_000, "RapidAPI")

# Hardware RK3588 (10 unités)
roi_hw, profit_hw = calculate_roi(5_000, 10_000, "Hardware RK3588 (10 unités)")

print("="*60)
print(f"\n✅ ROI Total Initiatives: €{profit_aws + profit_n8n + profit_prismatic + profit_ghl + profit_rapid + profit_hw:,}")
```
